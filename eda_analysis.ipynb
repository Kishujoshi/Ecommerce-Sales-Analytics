# Total Sales & Profit

```python
print ("Total Sales :", df["Sales"].sum())
print ("Total Profit :", df["Profit"].sum())
```

# Top 10 Customers

```python
Top_Customers = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending = False).head(10)
```

```python
Top_Customers
```

# Sales By Region & Visulization 

```python
Sales_Region = df.groupby("Region")["Sales"].sum().sort_values(ascending = False)
```

```python
Sales_Region
```

```python
plt.figure(figsize = (8,8))
sns.barplot(data = Sales_Region)
plt.title("Region Wise Sales")
plt.show()
```

# Category-wise Sales & Visulization

```python
category_wise_Sales = df.groupby("Category")["Sales"].sum()
```

```python
category_wise_Sales
```

```python
plt.figure(figsize = (8,8))
sns.barplot(data = category_wise_Sales)
plt.title("Category-wise Sales")
plt.show()
```

# Monthly Sales Trend & Visulization

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format = "%d-%m-%Y")
```

```python
monthly_sales = df.groupby(df["Order Date"].dt.month)["Sales"].sum()
```

```python
monthly_sales
```

```python
plt.figure(figsize = (12,8))
sns.barplot(data = monthly_sales)
plt.title("Monthly Sales Trend")
plt.show()
```

# Profit vs Discount

```python
plt.figure(figsize = (8,8))
sns.scatterplot(x = df["Discount"], y = df["Profit"])
plt.title("Profit vs Discount")
plt.show()
```

# Correlation Heatmap

```python
numeric_df = df[["Sales","Quantity", "Discount", "Profit"]]

plt.figure(figsize=(10,6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```
